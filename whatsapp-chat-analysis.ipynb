```python
import re
import pandas as pd
```

```python
f = open('WhatsApp Chat - Rupesh.txt','r',encoding='utf-8') #open text chat file using file handling
```

```python
data = f.read() #reading data in string format and store in variable
```

```python
print(data) # seperate the date and message
```

```python
print(type(data))
```

```python
# pattern = '\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s-\s'  # 23/12/24, 6:05:37
pattern = r'\[\d{2}/\d{2}/\d{2},\s*\d{1,2}:\d{2}:\d{2}\s*[AP]M\]\s*'
```

```python
messages = re.split(pattern,data)
print(messages)
len(messages)
```

```python
dates = re.findall(pattern,data)
dates
```

```python
messages = messages[:len(dates)]
```

```python
dates = [
    date.replace('[', '').replace(']', '').replace('\u202f', ' ').strip()
    for date in dates
]
```

```python
df = pd.DataFrame({'user_message':messages,'message_date':dates})
#convert message_date type
df['message_date'] = pd.to_datetime(
    df['message_date'],
    format='%d/%m/%y, %I:%M:%S %p'
)

# Rename column
df.rename(
    columns={'message_date': 'date'},
    inplace=True
)
df
```

```python
df.shape
```

```python
import re

users = []
messages = []

for message in df['user_message']:

    # If message is a list, convert it to string
    if isinstance(message, list):
        message = ' '.join(message)

    entry = re.split(r'^([^:]+):\s*', message, maxsplit=1)

    if len(entry) > 2:
        users.append(entry[1].strip())
        messages.append(entry[2].strip())
    else:
        users.append('group_notification')
        messages.append(entry[0].strip())

df['user'] = users
df['message'] = messages

df.drop(columns=['user_message'], inplace=True)

df.head()
```

```python
period = []
for hour in df[['day_name','hour']]['hour']:
    if hour == 23:
        period.append(str(hour) + "-" + str('00'))
    elif hour == 0:
        period.append(str('00') + "-" + str(hour+1))
    else:
        period.append(str(hour) + "-" + str(hour+1))
```

```python
df['period'] = period
df['period']
```

```python
import seaborn as sns
plt.Figure(figsize=(20,6))
sns.heatmap(df.pivot_table(index='day_name',columns='period',values='message',aggfunc='count').fillna(0))
plt.yticks(rotation='horizontal')
plt.show()
```
